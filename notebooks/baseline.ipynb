## Baseline 2 — Features couleur + Régression logistique

Chaque image est représentée par 30 variables : moyenne RGB (3), écart-type RGB (3) et histogramme RGB à 8 bins par canal (24). Les histogrammes sont normalisés afin de ne pas dépendre du nombre de pixels.

```python
def extract_features(image_path):
    """Extrait 30 features couleur d'une image RGB."""
    with Image.open(image_path) as image:
        pixels = np.asarray(image.convert("RGB"), dtype=np.float32) / 255.0

    pixels = pixels.reshape(-1, 3)
    means = pixels.mean(axis=0)
    stds = pixels.std(axis=0)

    histograms = []
    for channel in range(3):
        hist, _ = np.histogram(
            pixels[:, channel], bins=8, range=(0.0, 1.0)
        )
        histograms.append(hist / pixels.shape[0])

    features = np.concatenate([means, stds, *histograms]).astype(np.float32)
    assert features.shape == (30,)
    return features
```

```python
# Index unique des images des deux parties de HAM10000.
image_dirs = [
    PROJECT_ROOT / "data/raw/HAM10000_images_part_1",
    PROJECT_ROOT / "data/raw/HAM10000_images_part_2",
]
image_paths = {
    path.stem: path
    for image_dir in image_dirs
    for path in image_dir.glob("*.jpg")
}

missing_ids = set(train_val_df["image_id"]) - image_paths.keys()
if missing_ids:
    raise FileNotFoundError(
        f"{len(missing_ids)} image(s) introuvable(s), par exemple : {next(iter(missing_ids))}"
    )

print(f"{len(image_paths)} images indexées")
```

```python
# Fold 0 = validation ; folds 1 à 4 = entraînement.
train_df = train_val_df[train_val_df["fold"] != 0].copy()
val_df = train_val_df[train_val_df["fold"] == 0].copy()

def extract_features_for_dataframe(df):
    return np.vstack([
        extract_features(image_paths[image_id])
        for image_id in df["image_id"]
    ])

X_train = extract_features_for_dataframe(train_df)
X_val = extract_features_for_dataframe(val_df)
y_train = train_df["label_encoded"].to_numpy()
y_val = val_df["label_encoded"].to_numpy()

print("Train :", X_train.shape, y_train.shape)
print("Validation :", X_val.shape, y_val.shape)
```

```python
# La standardisation évite qu'un type de feature domine l'optimisation.
color_model = make_pipeline(
    StandardScaler(),
    LogisticRegression(max_iter=1000, random_state=42),
)
color_model.fit(X_train, y_train)
```

```python
val_predictions = color_model.predict(X_val)
color_metrics = compute_metrics(y_val, val_predictions)

pd.Series(color_metrics, name="baseline_couleur_logreg")
```
